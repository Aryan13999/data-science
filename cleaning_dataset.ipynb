```python
import json

def open_file(filename):
     with open (filename,'r') as f:
        data= json.load(f)
     return data

new_data=open_file('ds.json')
new_data
```

```python
def display_info(new_data):
    for info in new_data ['users']:
        print(f"{info['name']} friends: {info['friends']} liked pages :{info['liked_pages']}" )

for page in new_data['pages']:
    print(f"{page['name']}")



display_info(new_data)
```
